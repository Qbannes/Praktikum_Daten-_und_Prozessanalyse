```python
# Fehlerhafte Codes

lst = [17 , 12, 7, 34, 70, 8, 28, 55, 111]

# Berechnung der Summe
summe = 0
for element in lst:
    summe += element

# Berechnen des Durchschnitts
durchschnitt = summe / len(lst)
print(f"Durchschnitt: {durchschnitt},\n Summe: {summe}")
```

```python
# Liste mit Quadratzahlen
squares = []
for i in range(1 , 6):
    squares.append(i ** 2)
    print("Quadratzahlen:", squares)
print("-------------------------")
print("Quadratzahlen:", squares)
```

```python
# Liste mit Quadratzahlen mittels List Comprehension
squares1 = [i ** 2 for i in range(1,7)]
print(f'Quadratzahlen von 1-6: {squares1}')
```

```python
squares2 = [i ** 2 for i in range(1,15)    if i % 2 == 0                    ]
print(f'Quadratzahlen von 1-6: {squares2}')
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from io import StringIO

# CSV-Daten als String
csv_data = '''
;Paris;London;Berlin;Wien;Prag;Rom
2017;80000;48000;100000;88000;28000;70000
2018;88000;52000;110000;86000;29000;72000
2019;93000;57000;112000;90000;31000;62000
2020;82000;56000;111000;81000;44000;50000
2021;90000;57000;101000;71000;45000;43000
2022;102000;49000;104000;62000;49000;35000
2023;102000;42000;112000;59000;59000;45000
2024;109000;54000;121000;54000;48000;46000
'''

# Daten laden
df = pd.read_csv(StringIO(csv_data), sep=';', index_col=0)

years = [2021, 2022, 2023, 2024]

fig, axs = plt.subplots(2, 2, figsize=(12, 10))

for i, year in enumerate(years):
    ax = axs[i // 2, i % 2]
    data = df.loc[year]
    data.plot.pie(ax=ax, autopct='%1.1f%%', startangle=90, legend=False, ylabel='', title=str(year))

plt.tight_layout()
plt.show()
```
