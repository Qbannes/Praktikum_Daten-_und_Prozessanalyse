```python
from qiskit_aer import Aer
from qiskit_ibm_runtime import SamplerV2 as Sampler
from qiskit_algorithms import Grover, AmplificationProblem
from qiskit.circuit.library import PhaseOracle

import time
import matplotlib.pyplot as plt
```

```python
def optimized_linear_search(data, target):
    n = len(data)
    i = 0
    while i < n:
        if data[i] == target:
            return i
        i += 1
    return -1

def grover_search_qiskit(target_element, n_qubits=6):
    bitstring = format(target_element, f"0{n_qubits}b")
    expression = " & ".join([f"{'' if bit == '1' else '~'}a{i}" for i, bit in enumerate(bitstring)])
    oracle = PhaseOracle(expression)
    problem = AmplificationProblem(oracle)

    sampler = Sampler(backend=Aer.get_backend('aer_simulator'))
    grover = Grover(sampler=sampler)
    result = grover.amplify(problem)

    return result

sizes = [64]  # Datenbankgröße 2^6 = 64 Elemente
linear_times = []
grover_times = []
runs = 10

for size in sizes:
    data = list(range(size))
    target = size - 1  # Suche das letzte Element (Worst-Case)

    start = time.perf_counter()
    optimized_linear_search(data, target)
    linear_times.append(time.perf_counter() - start)

    start = time.perf_counter()
    result = grover_search_qiskit(target, n_qubits=6)
    grover_times.append(time.perf_counter() - start)

    print(f"Grover Ergebnis: {result.assignment}, Erfolg: {result.oracle_evaluation}")

plt.figure(figsize=(10,5))
plt.plot(sizes, linear_times, label='Optimierte lineare Suche O(N)', marker='o')
plt.plot(sizes, grover_times, label='Grover Algorithmus Qiskit', marker='o')
plt.xlabel('Datenbankgröße (N)')
plt.ylabel('Laufzeit (Sekunden)')
plt.title('Laufzeitvergleich: Lineare Suche vs. Grover Algorithmus (Qiskit)')
plt.legend()
plt.grid(True)
plt.show()
```
